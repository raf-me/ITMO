## 2.2 Проверка гипотез

### Подготовка данных

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("sample_data/california_housing_train.csv")

df["rooms_per_household"] = df["total_rooms"] / df["households"]
df["bedrooms_per_room"] = df["total_bedrooms"] / df["total_rooms"]
df["population_per_household"] = df["population"] / df["households"]

df.head()
```

### Гипотеза №1: "чем выше медианный доход района, тем выше медианная стоимость жилья"

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(data=df, x='median_income', y='median_house_value', alpha = 0.4)

plt.title('Зависимость медианной стоимости жилья от медианного дохода района')
plt.xlabel('Медианный доход района')
plt.ylabel('Медианная стоимость жилья')
plt.show()

corr = df["median_income"].corr(df["median_house_value"])
print(f"Коэффициент корреляции: {corr}")
```

**Вывод по гипотезе**

На графике видно, что при увеличении медианного дохода района медианная стоимость жилья также обычно возрастает. Корреляция между признаками положительная, поэтому гипотеза подтверждается. Медианный доход является одним из наиболее важных факторов, связанных со стоимостью жилья.

### Гипотеза №2: "Стоимость жилья зависит от географического положения района"

```python
plt.figure(figsize=(10, 8))

scatter = plt.scatter(df["longitude"], df["latitude"], c=df["median_house_value"], alpha=0.4, s=10)

plt.colorbar(scatter, label='Медианная стоимость жилья')
plt.title('Зависимость стоимости жилья от географического положения района')
plt.xlabel('Долгота')
plt.ylabel('Широта')
plt.show()

```

**Вывод по гипотезе**

На графике видно, что стоимость жилья распределена по территории Калифорнии неравномерно. В некоторых географических зонах наблюдается заметно более высокая медианная стоимость жилья. Следовательно, гипотеза подтверждается: географическое положение района связано со стоимостью жилья.

## Гипотеза №3: "Прибрежные районы Калифорнии имеют более высокую стоимость жилья"

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(data=df, x="longitude", y="median_house_value", alpha=0.4)

plt.title("Связь долготы и стоимости жилья")
plt.xlabel("Долгота")
plt.ylabel("Медианная стоимость жилья")
plt.show()

corr = df["longitude"].corr(df["median_house_value"])
print(f"Коэффициент корреляции: {corr:.3f}")
```

**Вывод по гипотезе**

На графике видно, что районы, расположенные западнее, чаще имеют более высокую стоимость жилья. Это можно связать с близостью к побережью и крупным городским агломерациям. Гипотеза в целом подтверждается, хотя одной только долготы недостаточно для точного определения близости к побережью.

## Гипотеза №9: В данных есть выбросы по численности населения

Предполагается, что в датасете есть территории с аномально высокой численностью населения. Такие значения могут отличаться от основной массы данных и влиять на результаты анализа.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("sample_data/california_housing_train.csv")

Q1 = df["population"].quantile(0.25)
Q3 = df["population"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers_population = df[
    (df["population"] < lower_bound) |
    (df["population"] > upper_bound)
]

print("Нижняя граница:", lower_bound)
print("Верхняя граница:", upper_bound)
print("Количество выбросов:", len(outliers_population))
print("Доля выбросов:", round(len(outliers_population) / len(df) * 100, 2), "%")
print("Максимальное значение population:", df["population"].max())

plt.figure(figsize=(10, 5))
sns.boxplot(x=df["population"])
plt.title("Выбросы по численности населения")
plt.xlabel("population")
plt.show()
```

**Вывод по гипотезе:**

В датасете присутствуют выбросы по численности населения. По критерию межквартильного размаха обнаружено 1015 выбросов (5.97% наблюдений). Максимальное значение численности населения составляет 35682 человека, что значительно превышает типичные значения выборки.

## Гипотеза №10. В данных есть выбросы по количеству комнат

Предполагается, что в датасете есть территории с аномально большим количеством комнат. Такие объекты могут отличаться от основной массы данных и влиять на анализ связи количества комнат со стоимостью жилья.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("sample_data/california_housing_train.csv")

Q1 = df["total_rooms"].quantile(0.25)
Q3 = df["total_rooms"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers_rooms = df[
    (df["total_rooms"] < lower_bound) |
    (df["total_rooms"] > upper_bound)
]

print("Нижняя граница:", lower_bound)
print("Верхняя граница:", upper_bound)
print("Количество выбросов:", len(outliers_rooms))
print("Доля выбросов:", round(len(outliers_rooms) / len(df) * 100, 2), "%")
print("Максимальное значение total_rooms:", df["total_rooms"].max())

plt.figure(figsize=(10, 5))
sns.boxplot(x=df["total_rooms"])
plt.title("Выбросы по общему количеству комнат")
plt.xlabel("total_rooms")
plt.show()
```

**Вывод по гипотезе**:

## Гипотеза №11. Стоимость жилья распределена неравномерно

Это свидетельствует о наличии верхнего ограничения стоимости жилья в датасете.

## Гипотеза №12. Медианный доход территорий распределён неравномерно

Предполагается, что медианный доход населения распределён неравномерно: большинство территорий имеет средний уровень дохода, а территорий с очень высоким доходом меньше.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("sample_data/california_housing_train.csv")

print(df["median_income"].describe())
print("Асимметрия распределения:", df["median_income"].skew())

plt.figure(figsize=(10, 5))
sns.histplot(df["median_income"], bins=40, kde=True)
plt.title("Распределение медианного дохода")
plt.xlabel("median_income")
plt.ylabel("Количество объектов")
plt.show()
```

## Гипотеза №13. Количество домохозяйств связано с количеством комнат и спален

Предполагается, что чем больше домохозяйств указано для территории, тем больше общее количество комнат и спален. Это логично, так как большее число домохозяйств обычно соответствует большему объёму жилья.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("sample_data/california_housing_train.csv")

correlation_table = df[[
    "households",
    "total_rooms",
    "total_bedrooms"
]].corr()

print(correlation_table)

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_table, annot=True, cmap="coolwarm")
plt.title("Связь количества домохозяйств, комнат и спален")
plt.show()

plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="households",
    y="total_rooms",
    alpha=0.3
)
plt.title("Связь количества домохозяйств и общего количества комнат")
plt.xlabel("households")
plt.ylabel("total_rooms")
plt.show()

plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="households",
    y="total_bedrooms",
    alpha=0.3
)
plt.title("Связь количества домохозяйств и количества спален")
plt.xlabel("households")
plt.ylabel("total_bedrooms")
plt.show()
```

**Вывод по гепотезе:**

Между количеством домохозяйств, количеством комнат и количеством спален наблюдается очень сильная положительная связь. Коэффициенты корреляции составляют 0.92 и 0.98 соответственно. Это означает, что увеличение числа домохозяйств сопровождается увеличением общего количества жилых помещений.

## Гипотеза №14. Плотность проживания влияет на стоимость жилья

Для проверки создаётся новый признак `population_per_household`, который показывает среднее количество жителей на одно домохозяйство. Предполагается, что более высокая плотность проживания может быть связана с более низкой стоимостью жилья.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("sample_data/california_housing_train.csv")

df["population_per_household"] = df["population"] / df["households"]

correlation = df["population_per_household"].corr(df["median_house_value"])

print("Корреляция population_per_household и median_house_value:", correlation)
print(df["population_per_household"].describe())

plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="population_per_household",
    y="median_house_value",
    alpha=0.3
)
plt.title("Связь плотности проживания и стоимости жилья")
plt.xlabel("population_per_household")
plt.ylabel("median_house_value")
plt.show()

df["population_density_group"] = pd.qcut(
    df["population_per_household"],
    q=4,
    duplicates="drop"
)

group_table = df.groupby("population_density_group")["median_house_value"].median()

print(group_table)

plt.figure(figsize=(10, 5))
group_table.plot(kind="bar")
plt.title("Медианная стоимость жилья по группам плотности проживания")
plt.xlabel("Группа по population_per_household")
plt.ylabel("Медианная стоимость жилья")
plt.xticks(rotation=45)
plt.show()
```

**Вывод по гепотезе:**

На диаграмме рассеяния линейная зависимость между плотностью проживания и стоимостью жилья выражена слабо. Однако группировка данных показывает устойчивое снижение медианной стоимости жилья при увеличении количества жителей на одно домохозяйство: от 221850 до 152050 долларов. Следовательно, более высокая плотность проживания связана с более низкой стоимостью жилья.

## Гипотеза №15. Чем больше комнат приходится на одно домохозяйство, тем выше стоимость жилья

Для проверки создаётся новый признак `rooms_per_household`, который показывает среднее количество комнат на одно домохозяйство. Предполагается, что большее количество комнат на домохозяйство связано с более высокой стоимостью жилья.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("sample_data/california_housing_train.csv")

df["rooms_per_household"] = df["total_rooms"] / df["households"]

correlation = df["rooms_per_household"].corr(df["median_house_value"])

print("Корреляция rooms_per_household и median_house_value:", correlation)
print(df["rooms_per_household"].describe())

plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="rooms_per_household",
    y="median_house_value",
    alpha=0.3
)
plt.title("Связь комнат на домохозяйство и стоимости жилья")
plt.xlabel("rooms_per_household")
plt.ylabel("median_house_value")
plt.show()

df["rooms_group"] = pd.qcut(
    df["rooms_per_household"],
    q=4,
    duplicates="drop"
)

group_table = df.groupby("rooms_group")["median_house_value"].median()

print(group_table)

plt.figure(figsize=(10, 5))
group_table.plot(kind="bar")
plt.title("Медианная стоимость жилья по группам rooms_per_household")
plt.xlabel("Группа по rooms_per_household")
plt.ylabel("Медианная стоимость жилья")
plt.xticks(rotation=45)
plt.show()
```

**Вывод по гипотезе:**

Анализ показал, что при увеличении среднего количества жителей на одно домохозяйство медианная стоимость жилья уменьшается. Для территорий с наименьшей плотностью проживания медианная стоимость жилья составляет 221850 долларов, тогда как для территорий с наибольшей плотностью — 152050 долларов. Таким образом, более высокая плотность проживания связана с более низкой стоимостью жилья.
